**contexto**

La empresa quiere predecir el éxito de futuras aplicaciones móviles. Han extraído un volcado de datos de la Google Play Store, pero el pipeline de Machine Learning rechaza el archivo. Su misión como ingenieros de datos es sanear los valores faltantes aplicando reglas de negocio estrictas.

```python
import pandas as pd
import numpy as np
 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)
```

```python
df.isnull().sum()
```

*Ejecuten df.isnull().sum(). Identificarán que la columna Rating tiene casi 1,500 valores nulos, mientras que Type y Android Ver tienen solo 1 o 2.*

### **Pregunta A: ¿Por qué es arquitectónicamente seguro eliminar las filas donde falta Android Ver, pero sería un error fatal eliminar las filas donde falta Rating?**

por el simple hecho que la magnitud es completamente diferente entre borrar solo 3 filas de informacion a borrar casi una 1/10 parte

```python
print(len(df))
df = df.dropna(subset=['Android Ver', 'Type'])
print(len(df))
```

### **Pregunta B: Verifiquen que el total de registros en el DataFrame apenas disminuyó**

pasó de 10841 a 10837, osea si

```python
medianas = df.groupby('Category')['Rating'].transform('median')
df['Rating'] = df['Rating'].fillna(medianas)

df['Rating'].isnull().sum()
```

*Rellenar los 1,500 nulos de Rating con el promedio global es un error, ya que las apps de la categoría FAMILY (Familia) no se califican igual que las de DATING (Citas).*

### **Pregunta C: Utilizando la lógica de la semana pasada, escriban un bloque combinando .groupby('Category') y .transform() para rellenar los NaN de la columna Rating usando la mediana de su propia categoría.**

```python
df['Size'] = df['Size'].replace('Varies with device', np.nan)

df['Size'].isnull().sum()
```
