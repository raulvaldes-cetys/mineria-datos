# Técnicas de sumarización — Titanic

Asimetría, curtosis y tablas cruzadas.

```python
import pandas as pd


url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

print(df.head())
```

```python
df["Age"].skew()
```

### 1. Asimetría (Skewness) de `Age`

**Resultado: +0.389** — asimetría positiva.

**Qué significa:** la cola larga está del lado derecho, o sea del lado de las edades altas. Había más pasajeros jóvenes que ancianos: la mayoría se concentra entre los 20 y 35 años, y los pocos adultos mayores estiran la distribución hacia la derecha (solo 22 pasajeros pasaban de 60 años, contra 384 menores de 30).

Se confirma con media (29.70) > mediana (28.0): los mayores jalan el promedio hacia arriba. Al ser solo +0.39, la distribución es casi simétrica.

```python
df["Fare"].kurt()
```

### 2. Curtosis (Kurtosis) de `Fare`

**Resultado: +33.40** — curtosis muy alta.

**Qué significa:** Un valor de 33 indica colas extremadamente gruesas, osea que hay muchísimos más valores lejos del centro de los que una normal predeciría, y al mismo tiempo un pico muy pronunciado de tarifas baratas.

Sí confirma los outliers. Hay los 116 pasajeros por encima de 65.63 y el caso extremo de 512.33 son precisamente lo que infla la curtosis a 33. Una distribución normal no produce un numero asi

```python
pd.crosstab(df["Pclass"], df["Survived"])
```

```python
pd.crosstab(df["Pclass"], df["Survived"], normalize="index")
```

### 3. Tabla cruzada (Cross-Tabulation) `Pclass` × `Survived`

| Pclass | Murió (0) | Sobrevivió (1) | % sobrevivió |
|---|---|---|---|
| 1ª | 80 | 136 | **62.96%** |
| 2ª | 97 | 87 | 47.28% |
| 3ª | 372 | 119 | **24.24%** |

**Patrón visible:** la supervivencia cae de forma monótona conforme baja la clase. La primera clase es el único grupo donde sobrevivieron más de los que murieron, en tercera murieron más de 3 por cada sobreviviente.

En números crudos el contraste es aún más fuerte, los 372 muertos de tercera clase son, por sí solos, más que todos los pasajeros de primera juntos. La clase socioeconómica funcionó como predictor directo de supervivencia.
